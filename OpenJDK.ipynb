```python
# ✅ Clone OpenJDK, fetch PR #25872, and get diffs between merge base and PR tip

# STEP 1: Clone repo shallowly
!git clone --depth=1 https://github.com/openjdk/jdk.git
%cd jdk

# STEP 2: Fetch the PR branch from GitHub
!git fetch origin pull/25872/head:pr-25872

# STEP 3: Checkout to that PR branch
!git checkout pr-25872

# STEP 4: Identify GOOD and BAD SHAs
GOOD_SHA = !git merge-base origin/master pr-25872
GOOD_SHA = GOOD_SHA[0]

BAD_SHA = !git rev-parse HEAD
BAD_SHA = BAD_SHA[0]

print("✅ GOOD SHA:", GOOD_SHA)
print("❌ BAD SHA:", BAD_SHA)

# STEP 5: Get commit messages and changed files between GOOD and BAD
log_output = !git log --oneline {GOOD_SHA}..{BAD_SHA}
diff_output = !git diff --name-status {GOOD_SHA} {BAD_SHA}

print("\n=== Commits Between Builds ===")
for line in log_output:
    print(line)

print("\n=== Changed Files (Name Status) ===")
for line in diff_output:
    print(line)
```
